## Step 4: Upload Investigation Skill

```python
%%bash
cd /workshop/deployment/devops-agent/skills/sql-server-investigation
zip -r ../sql-server-investigation.zip SKILL.md references/
echo "✅ Skill zip created:"
ls -la /workshop/deployment/devops-agent/skills/sql-server-investigation.zip
```

### Upload via Console

1. Open the [DevOps Agent console](https://console.aws.amazon.com/aidevops/home#/agent-spaces)
2. Click **sql-server-dbops** → **Operator access** → **Skills** → **Add skill** → **Upload skill**
3. Select `sql-server-investigation.zip`, Agent Type = **Generic**, click **Upload**

## Key Takeaways

- **AgentCore Gateway** exposes any Lambda as MCP tools (zero agent code)
- **IAM auth (SigV4)** — no OAuth tokens or Cognito setup needed
- **DevOps Agent** handles reasoning and orchestration — you just provide tools
- **Skills** teach the agent domain-specific investigation workflows
- **Same tools** work from CLI agents, Kiro, and DevOps Agent
